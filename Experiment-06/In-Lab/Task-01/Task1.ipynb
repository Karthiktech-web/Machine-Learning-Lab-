```python
# ============================================================
# EXPERIMENT 06 - TASK 01
# Principal Component Analysis (PCA)
#
# Name    : R. Karthik
# Roll No : 25EU02904
# ============================================================

import warnings
warnings.filterwarnings("ignore")

import numpy as np
import pandas as pd
import matplotlib.pyplot as plt

from sklearn.datasets import load_iris
from sklearn.preprocessing import StandardScaler
from sklearn.decomposition import PCA

print("=" * 60)
print("EXPERIMENT 06 - TASK 01")
print("Principal Component Analysis")
print("Name    : R. Karthik")
print("Roll No : 25EU02904")
print("=" * 60)

# ------------------------------------------------------------
# 1. Load Iris Dataset
# ------------------------------------------------------------

iris = load_iris()

X = iris.data
y = iris.target

feature_names = iris.feature_names
target_names = iris.target_names

print("\nDataset Shape:", X.shape)

print("\nFeatures:")
print(feature_names)

print("\nClasses:")
print(target_names)

# ------------------------------------------------------------
# 2. Standardize Features
# ------------------------------------------------------------

scaler = StandardScaler()

X_scaled = scaler.fit_transform(X)

print("\nStandardized Dataset Shape:")
print(X_scaled.shape)

# ------------------------------------------------------------
# 3. Covariance Matrix
# ------------------------------------------------------------

cov_matrix = np.cov(
    X_scaled.T
)

print("\nCovariance Matrix:")
print(
    np.round(cov_matrix, 4)
)

# ------------------------------------------------------------
# 4. Eigenvalue Decomposition
# ------------------------------------------------------------

eigenvalues, eigenvectors = np.linalg.eig(
    cov_matrix
)

print("\nEigenvalues:")
print(
    np.round(eigenvalues, 4)
)

# ------------------------------------------------------------
# 5. Apply PCA
# ------------------------------------------------------------

pca = PCA(
    n_components=2
)

X_pca = pca.fit_transform(
    X_scaled
)

print("\nPCA Transformed Shape:")
print(X_pca.shape)

pca_df = pd.DataFrame(
    X_pca,
    columns=[
        "PC1",
        "PC2"
    ]
)

pca_df["Species"] = [
    target_names[i]
    for i in y
]

print("\nFirst Five PCA Rows:")
print(
    pca_df.head()
)

# ------------------------------------------------------------
# 6. Explained Variance
# ------------------------------------------------------------

explained_variance = (
    pca.explained_variance_ratio_
)

print("\nExplained Variance Ratio:")

for i, value in enumerate(
    explained_variance,
    start=1
):
    print(
        f"PC{i}: {value * 100:.2f}%"
    )

total_variance = (
    explained_variance.sum()
)

print(
    f"\nTotal Variance Retained: "
    f"{total_variance * 100:.2f}%"
)

# ------------------------------------------------------------
# 7. PCA Visualization
# ------------------------------------------------------------

plt.figure(
    figsize=(8, 6)
)

for class_id, class_name in enumerate(
    target_names
):

    mask = y == class_id

    plt.scatter(
        X_pca[mask, 0],
        X_pca[mask, 1],
        label=class_name
    )

plt.xlabel("Principal Component 1")

plt.ylabel("Principal Component 2")

plt.title(
    "PCA Projection of Iris Dataset"
)

plt.legend()

plt.grid(True)

plt.show()

# ------------------------------------------------------------
# Conclusion
# ------------------------------------------------------------

print("\n" + "=" * 60)
print("TASK 01 COMPLETED SUCCESSFULLY")
print("PCA reduced the Iris dataset from 4 dimensions to 2 dimensions.")
print("Name    : R. Karthik")
print("Roll No : 25EU02904")
print("=" * 60)
```
