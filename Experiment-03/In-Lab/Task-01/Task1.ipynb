# Task 1: Supervised Classification on Iris Dataset

Implement Logistic Regression, Linear Discriminant Analysis (LDA), and k-Nearest Neighbors (k-NN) classifiers and compare their performance.

```python
import pandas as pd
```

```python
from sklearn.datasets import load_iris
```

```python
from sklearn.model_selection import train_test_split
```

```python
from sklearn.preprocessing import StandardScaler
```

```python
import warnings
```

```python
warnings.filterwarnings("ignore")
```

```python
iris_raw = load_iris()
```

```python
X_iris = iris_raw.data
```

```python
y_iris = iris_raw.target
```

```python
scaler_iris = StandardScaler()
```

```python
X_iris_scaled = scaler_iris.fit_transform(X_iris)
```

```python
X_train, X_test, y_train, y_test = train_test_split(X_iris_scaled, y_iris, test_size=0.20, stratify=y_iris, random_state=42)
```

```python
print("Training set shape:", X_train.shape)
```

```python
print("Test set shape:", X_test.shape)
```

```python
from sklearn.linear_model import LogisticRegression
```

```python
from sklearn.metrics import accuracy_score
```

```python
lr_model = LogisticRegression(multi_class="multinomial", solver="lbfgs", random_state=42)
```

```python
lr_model.fit(X_train, y_train)
```

```python
lr_preds = lr_model.predict(X_test)
```

```python
lr_accuracy = accuracy_score(y_test, lr_preds)
```

```python
print(f"Logistic Regression Test Accuracy: {lr_accuracy * 100:.2f}%")
```

```python
from sklearn.discriminant_analysis import LinearDiscriminantAnalysis
```

```python
lda_model = LinearDiscriminantAnalysis()
```

```python
lda_model.fit(X_train, y_train)
```

```python
lda_preds = lda_model.predict(X_test)
```

```python
lda_accuracy = accuracy_score(y_test, lda_preds)
```

```python
print(f"LDA Test Accuracy: {lda_accuracy * 100:.2f}%")
```

```python
from sklearn.neighbors import KNeighborsClassifier
```

```python
knn = KNeighborsClassifier()
```

```python
knn.fit(X_train, y_train)
```

```python
y_pred_knn = knn.predict(X_test)
```

```python
knn_accuracy = accuracy_score(y_test, y_pred_knn)
```

```python
print("k-NN Classification Accuracy:", knn_accuracy)
```

```python
comparison_df = pd.DataFrame({
    "Supervised Classifier": [
        "Logistic Regression",
        "Linear Discriminant Analysis (LDA)",
        "k-Nearest Neighbors (k-NN)"
    ],
    "Test Accuracy (%)": [
        lr_accuracy * 100,
        lda_accuracy * 100,
        knn_accuracy * 100
    ]
})
```

```python
print("--- Performance Comparison on Iris Dataset ---")
```

```python
print(comparison_df.to_string(index=False))
```

## Conclusion

The Iris dataset was successfully classified using Logistic Regression, Linear Discriminant Analysis, and k-Nearest Neighbors. The models were evaluated using test accuracy. The comparison shows the effectiveness of different supervised classification approaches on the Iris dataset.
